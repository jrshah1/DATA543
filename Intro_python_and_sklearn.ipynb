# 2 Python fundamentals (variables, functions, for-loops)

## Variables

Python is dynamically typed: variables can hold different types of values, types are inferred automatically.

```python
# Variables (types are inferred automatically)
ok = True        # bool
a = 5            # int
b = 2.5          # float
name = "risk"    # str
num_list = [False,1,2.7,'3']   # list

print("a =", a, "| type:", type(a))
print("b =", b, "| type:", type(b))
print("name =", name, "| type:", type(name))
print("ok =", ok, "| type:", type(ok))
print("num_list =", num_list, "| type:", type(num_list))
```

## Functions

A **function** is a reusable block of code that performs a specific task.

Why functions matter:
- Avoid repeating the same code
- Make programs easier to read
- Break complex logic into small, understandable pieces

A function:
- Takes **inputs** (arguments)
- Performs some computation
- Returns an **output**


```python
def square(x):
    return x * x

print("square(3) =", square(3))
print("square(10) =", square(10))
```

## For-loops

Loops repeat operations.


```python
nums = [1, 2, 3, 4, 5]

total = 0
for x in nums:
    total += x

print("nums =", nums)
print("sum(nums) computed by loop =", total)
```

# 3 Packages and imports (`time` example)

Before using a package, you must **import** it.
If you try to use a function without importing its package, Python will raise an error.

We will time a simple loop to measure how long it takes.


```python
import time

start = time.time()

s = 0
for i in range(100):  # 100
    s += i

end = time.time()

print("s =", s)
print("Elapsed seconds:", end - start)
```

# 4 NumPy

**NumPy** is the core numerical library used in machine learning and data science.

Why not Python lists?
- NumPy arrays are **much faster**
- They use **less memory**
- They are designed for **mathematical operations**

One of the key advantages is **vectorization**:
- Operations apply to **entire arrays at once**
- No explicit Python `for` loops are needed
- This leads to simpler code and major speedups

We will compare:
- Python loop (slow)
- NumPy vectorized operation (fast)


```python
import numpy as np
# We do not need to import time again as we already import it above

N = 2000000
```

Loop version

```python
out_loop = [1] * N
print(out_loop[:100])
```

```python
start = time.time()
for i in range(N):
    out_loop[i] = out_loop[i] * 2
t_loop = time.time() - start
print(out_loop[:100])
print(f"Loop time: {t_loop:.4f} seconds")
```

```python
# --- NumPy vectorized version ---
x = np.ones(N)  # [0,1,2,...,N-1]
start = time.time()
out_vec = x * 2
t_vec = time.time() - start

print(f"N = {N:,}")
print(f"Loop time: {t_loop:.4f} seconds")
print(f"NumPy vectorized time: {t_vec:.4f} seconds")
print("Speedup (loop / numpy):", t_loop / t_vec if t_vec > 0 else float("inf"))
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 50)
y_true = 2.0 * x + 1.0
y_noisy = y_true + np.random.randn(len(x)) * 2.0

# --- Subplots: scatter + line ---
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

# Left: scatter
ax[0].scatter(x, y_noisy)
ax[0].set_title("Scatter (noisy points)")
ax[0].set_xlabel("x")
ax[0].set_ylabel("y")

# Right: scatter + line
ax[1].scatter(x, y_noisy, label="Noisy data")
ax[1].plot(x, y_true, label="True line", linewidth=3)
ax[1].set_title("Scatter + line")
ax[1].set_xlabel("x")
ax[1].set_ylabel("y")
ax[1].legend()

plt.show()
```

# 6 Sklearn


Scikit-Learn is a Python package designed to give access to well-known machine learning algorithms within Python code, through a clean, well-thought-out API.

It has been built by hundreds of contributors from around the world, and is used across industry and academia.

## Setup

```python
# Setup: imports used across all three tasks
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

# For reproducibility in train/test splits and some models
RANDOM_STATE = 0
```

## Classification (Supervised): Iris → Logistic Regression

**Problem:** Predict which *species* an iris flower belongs to (3 classes).  
**Inputs (features):** 4 measurements per flower.  
**Output (label):** species ID (0/1/2).

We'll report:
- **Accuracy**: fraction correct on the test set
- **Confusion matrix**: where mistakes happen


```python
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay

# Load dataset
X, y = load_iris(return_X_y=True)

target_names = load_iris().target_names

for i in range(0,150,35):
    print(f"Sample {i}:")
    print("  Features:", X[i])
    print("  Label:", y[i], f"({target_names[y[i]]})")
    print()
```

```python
# Train/test split (stratify keeps class balance)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=RANDOM_STATE, stratify=y
)

# Pipeline = standardize features + classifier
clf = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000)
)

clf.fit(X_train, y_train)
pred = clf.predict(X_test)

print("Classification accuracy:", accuracy_score(y_test, pred))

cm = confusion_matrix(y_test, pred)
ConfusionMatrixDisplay(cm).plot()
plt.title("Classification: Iris (Logistic Regression)")
plt.show()
```

## Regression (Supervised): Diabetes → Ridge Regression

**Problem:** Predict a **continuous** target **y** from a single input feature **x**.

**Key difference from classification:** output is a number, not a category.

We'll report:
- **Slope** and **Intercept**.
- **MSE** (mean squared error): smaller is better


```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Generate simple 1D data
np.random.seed(0)
X = np.linspace(0, 10, 50).reshape(-1, 1)   # shape (n_samples, 1)
y = 2.5 * X.squeeze() + 1.0 + np.random.randn(50) * 2.0

# Fit linear regression
model = LinearRegression()
model.fit(X, y)

# Predict on a dense grid for smooth line
X_line = np.linspace(0, 10, 100).reshape(-1, 1)
y_line = model.predict(X_line)

# Plot data points and regression line
plt.scatter(X, y, label="Data points")
plt.plot(X_line, y_line, label="Regression line", color='orange')
plt.xlabel("x")
plt.ylabel("y")
plt.title("1D Linear Regression y=ax+b")
plt.legend()
plt.show()

# Print learned parameters
print("Slope (a):", model.coef_[0])
print("Intercept (b):", model.intercept_)

# MSE
mse = mean_squared_error(y, model.predict(X))
print("Mean Squared Error (MSE):", mse)
```

## Clustering (Unsupervised): Synthetic blobs → KMeans

**Problem:** Given points **without labels**, find groups (clusters).  
**Unsupervised learning:** there is no “correct label” provided to the algorithm.

We'll:
- generate a simple 2D dataset (`make_blobs`)
- run **KMeans** to assign a cluster ID to each point
- plot the clusters and centers


```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

X, _ = make_blobs(
    n_samples=400, centers=3, random_state=RANDOM_STATE, cluster_std=0.7
)

plt.scatter(X[:, 0], X[:, 1])
plt.title("Synthetic blobs")
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=RANDOM_STATE, n_init=10)
labels = kmeans.fit_predict(X)

plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.scatter(
    kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
    marker="x", s=200
)
plt.title("Clustering: KMeans on synthetic blobs")
plt.show()
```
